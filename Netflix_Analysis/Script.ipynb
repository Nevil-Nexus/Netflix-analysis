Top 10 Most
common Genres

```python
data['genres']=data['genres'].apply(lambda x: x.split(','))
genre_counts=pd.Series(sum(data['genres'],[])).value_counts().head(10)
genre_counts
```

```python
plt.figure(figsize=(10, 6))
sns.barplot(x=genre_counts.values, y=genre_counts.index,
            hue=genre_counts.index, palette='cool', legend=False)
plt.title('Top 10 des genres les plus fréquents')
plt.xlabel('Nombre de titres')  
plt.ylabel('Genres')
plt.show()
```

Répartition des films selon leur durée

```python
data['duration_num'] = (
    data['duration']
    .astype('string')
    .str.extract(r'(\d+)')[0]
    .astype(float)
)

Duration = data['duration_num']
Duration
```

```python
plt.figure(figsize=(10,6))
sns.histplot(data[data['type']=='Movie'],x='duration_num',bins=30,kde=True,color='blue')
plt.title('Répartition de la durée des films')
plt.xlabel('Durée (minutes)')
plt.ylabel('Frequence')
plt.show()
```

Top 10 des années avec le plus de films et séries sortis

```python
titles_per_year=data.groupby('release_year').size()
titles_per_year
```

```python
top_10_years=titles_per_year.sort_values(ascending=False).head(10)
top_10_years
```

```python
plt.figure(figsize=(10,6))
sns.lineplot(x=top_10_years.index,y=top_10_years.values,marker='o',color='green')
plt.title('Top 10 des années avec le plus de titres sortis', fontsize=16)
plt.xlabel('Année')
plt.ylabel('Nombre de titres',fontsize=14)
plt.show()
```
